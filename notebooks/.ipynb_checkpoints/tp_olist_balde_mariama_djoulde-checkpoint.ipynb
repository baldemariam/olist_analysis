Analyse de l’E-Commerce Brésilien (Olist)

L'objectif:c'est d'nalyser les données e-commerce pour extraire des insights business.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Charger les doonnées(dataset Olist)

```python
orders = pd.read_csv("olist_orders_dataset.csv")
items = pd.read_csv("olist_order_items_dataset.csv")
products = pd.read_csv("olist_products_dataset.csv")
customers = pd.read_csv("olist_customers_dataset.csv")
payments = pd.read_csv("olist_order_payments_dataset.csv")
```

Explorer les données

```python
orders.head()
orders.info()
orders.describe()
```

```python
#Fusionner les données
```

```python
#=
```

```python
orders.isnull().sum()
```

```python
#Les données contiennent quelques valeurs manquantes, mais elles sont gérables.
```

```python
#FUSION (IMPORTANT 🔥)
```

```python
df = orders.merge(items, on="order_id")
df = df.merge(products, on="product_id")
df = df.merge(customers, on="customer_id")
df = df.merge(payments, on="order_id")
```

```python
top_items = df.groupby("item_name").size().sort_values(ascending=False)
print(top_items.head(10))
```

#Revenu par produit

```python
df["payment_value"].sum()
```

#Le chiffre d’affaires total montre l’importance du volume des transactions.
📦 Produits les plus vendus

```python
df["product_id"].value_counts().head(10)
```

#📅 Commandes par mois

```python
df["order_purchase_timestamp"] = pd.to_datetime(df["order_purchase_timestamp"])
df["month"] = df["order_purchase_timestamp"].dt.month
df["month"].value_counts().sort_index()
```

#🟢 VISUALISATION

```python
df["month"].value_counts().sort_index().plot(kind="bar")
plt.title("Commandes par mois")
plt.show()
```

#🟢 INTERPRÉTATION (TRÈS IMPORTANT)

Certains mois ont plus de commandes (effet saisonnier)
- Les produits populaires dominent les ventes
- Le chiffre d’affaires est concentré sur certains segments
